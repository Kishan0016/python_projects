# Count how many times a number appears in a list.

```python
numbers = [1, 2, 1, 3, 2, 1, 3, 3, 3, 3]

count_dict = {}

for num in numbers:
    if num in count_dict:
        count_dict[num] += 1
    else:
        count_dict[num] = 1

print(count_dict)

    
```

# Write a function to check whether a number is prime

```python
def prime():
    num = int(input("Enter a number: "))
    
    if num > 1:
        for i in range(2, num):
            if num % i == 0:
                print("Not a Prime Number")
                break
        else:
            print("Prime Number")
    else:
        print("Not a Prime Number")
prime()
```

# Write a function to find the factorial of a numbe

```python
def fact():
    num = int(input("Enter the number:- "))
    fact = 1
    for i in range(1, num+1):
        fact *= i
    print(fact)
fact()
```

# Create a dictionary of student names and marks and print them

```python
def student():
    data = {}
    stu = int(input("How Many student is there:- "))
    for i in range(1, stu+1):
        name = input(f"Enter the name of student {i}:- ").lower()
        maks = int(input(f"Enter the marks of {name}:- "))

        data[name] = maks
    print(data)
        
student()
```

# Find the student with the highest marks

```python
students = {
    "Aman": 78,
    "Riya": 85,
    "Neha": 92,
    "Rahul": 88
}

top_student = max(students, key=students.get)

print("Student with highest marks:", top_student)
print("Marks:", students[top_student])
```
