# TO-DO LIST

```python
# let's create a to-do list
```

```python
try:
    name = input("Enter Your Name Here:- ")
    password = input("Please Enter The Password:- ")

    if password == 'Admin':

        print('-' * 45)
        print(' ' * 10, 'Welcome To TO DO LIST App')
        print('-' * 45)

        work = []

        def show_list():
            if not work:
                print("Your To-Do List is Empty.")
            else:
                print("\nYour Tasks:")
                for i, task in enumerate(work, start=1):
                    print(f"{i}. {task}")

        def add_work():
            task = input("Enter New Task: ")
            work.append(task)
            print("Task Added Successfully.")

        def work_done():
            show_list()
            if work:
                num = int(input("Enter Task Number Completed: "))
                if 1 <= num <= len(work):
                    print(f"Task Completed: {work[num-1]}")
                else:
                    print("Invalid Task Number.")

        def remove_work():
            show_list()
            if work:
                num = int(input("Enter Task Number To Remove: "))
                if 1 <= num <= len(work):
                    removed = work.pop(num - 1)
                    print(f"Removed Task: {removed}")
                else:
                    print("Invalid Task Number.")

        while True:
            print("\nChoose Any One Option")
            print("1. Show The List")
            print("2. Work Done")
            print("3. Add Work")
            print("4. Remove Work")
            print("5. Exit")

            choice = input("You Want To Continue (y/n): ").lower()
            if choice != 'y':
                print(f"Thank You! For Using {name}")
                break

            option = input(f"Enter Your Choice {name}: ")

            if option == '1':
                show_list()
            elif option == '2':
                work_done()
            elif option == '3':
                add_work()
            elif option == '4':
                remove_work()
            elif option == '5':
                print(f"Thank You! For Using {name}")
                break
            else:
                print("Please Choose A Valid Number (1–5)")

    else:
        print("Wrong Password! Please Re-check Your Password!")

except Exception as Err:
    print("Error:", Err)
```
